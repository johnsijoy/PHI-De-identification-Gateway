```python
# DOB AND AGE

import re
from datetime import datetime

def extract_dob_and_age(note):
    dob_match = re.search(
        r"DOB\s+(\d{2}-\d{2}-\d{4})",
        str(note)
    )

    age_match = re.search(
        r"age\s+(\d+)",
        str(note),
        re.IGNORECASE
    )

    visit_match = re.search(
        r"on\s+(\d{2}-\d{2}-\d{4})",
        str(note)
    )

    if not dob_match or not age_match or not visit_match:
        return None

    dob = datetime.strptime(dob_match.group(1), "%d-%m-%Y")
    visit_date = datetime.strptime(visit_match.group(1), "%d-%m-%Y")
    stated_age = int(age_match.group(1))

    calculated_age = (
        visit_date.year - dob.year
        - ((visit_date.month, visit_date.day) < (dob.month, dob.day))
    )

    return stated_age, calculated_age


for dataset_name, dataset in [
    ("TRAINING", training_df),
    ("TESTING", testing_df)
]:

    print(dataset_name, "AGE CONSISTENCY CHECK")


    mismatch_count = 0

    for _, row in dataset.iterrows():
        result = extract_dob_and_age(row["Clinical note"])

        if result is not None:
            stated_age, calculated_age = result

            if stated_age != calculated_age:
                mismatch_count += 1
                print(
                    row["Note ID"],
                    "| Stated age:", stated_age,
                    "| Calculated age:", calculated_age
                )

    print("\nTotal age mismatches:", mismatch_count)
```

```python
# CORRECT DOB AND AGE MISMATCHES

import re
from datetime import datetime

def calculate_correct_age(note):

    dob_match = re.search(
        r"DOB\s+(\d{2}-\d{2}-\d{4})",
        str(note)
    )

    visit_match = re.search(
        r"\bon\s+(\d{2}-\d{2}-\d{4})",
        str(note)
    )

    if not dob_match or not visit_match:
        raise ValueError("DOB or visit date not found in clinical note.")

    dob = datetime.strptime(dob_match.group(1), "%d-%m-%Y")
    visit_date = datetime.strptime(visit_match.group(1), "%d-%m-%Y")

    age = visit_date.year - dob.year

    if (visit_date.month, visit_date.day) < (dob.month, dob.day):
        age -= 1

    return age


def correct_dataset_ages(dataset):

    dataset = dataset.copy()

    for index, row in dataset.iterrows():

        note = str(row["Clinical note"])
        correct_age = calculate_correct_age(note)

        # Update age inside the clinical note
        updated_note = re.sub(
            r"(\bage\s+)\d+",
            lambda match: match.group(1) + str(correct_age),
            note,
            count=1,
            flags=re.IGNORECASE
        )

        dataset.at[index, "Clinical note"] = updated_note

        # Update the structured Patient Age column
        dataset.at[index, "Patient Age"] = correct_age

    return dataset


# Correct both datasets
training_df = correct_dataset_ages(training_df)
testing_df = correct_dataset_ages(testing_df)


# Verify corrected ages
for dataset_name, dataset in [
    ("TRAINING", training_df),
    ("TESTING", testing_df)
]:

    mismatch_count = 0

    for _, row in dataset.iterrows():

        calculated_age = calculate_correct_age(row["Clinical note"])
        stated_age = int(row["Patient Age"])

        if calculated_age != stated_age:
            mismatch_count += 1

    print(dataset_name, "remaining mismatches:", mismatch_count)


# Save corrected datasets
training_df.to_csv("PHI_Training_Dataset.csv", index=False)
testing_df.to_csv("PHI_Testing_Dataset.csv", index=False)

print("\nSaved successfully.")
```

# **PHI ANNOTATTION** - Training

```python

# PHI ANNOTATION

import re
import json

# PHI CATEGORIES

PHI_LABELS = [
    "PATIENT_NAME",
    "RELATIVE_NAME",
    "CLINICIAN_NAME",
    "FACILITY",
    "LOCATION",
    "DATE",
    "AGE_OVER_89",
    "PHONE",
    "FAX",
    "EMAIL",
    "SSN",
    "MRN",
    "HEALTH_PLAN_ID",
    "ACCOUNT_NUMBER",
    "LICENSE_NUMBER",
    "HOSPITAL_ID",
    "URL",
    "IP_ADDRESS",
    "VEHICLE_ID",
    "VEHICLE_SERIAL",
    "LICENSE_PLATE",
    "DEVICE_ID",
    "BIOMETRIC_ID",
    "PHOTO_ID",
    "UNIQUE_ID"
]


#  MAP DATASET COLUMNS TO PHI CATEGORIES

COLUMN_LABELS = {
    "Patient Name": "PATIENT_NAME",
    "Doctor Name": "CLINICIAN_NAME",
    "Relatives Names": "RELATIVE_NAME",
    "Hospital Name": "FACILITY",
    "Address": "LOCATION",
    "ZIP Code": "LOCATION",
    "Patient SSN": "SSN",
    "MRN": "MRN",
    "Health Plan ID": "HEALTH_PLAN_ID",
    "Account No": "ACCOUNT_NUMBER",
    "License No": "LICENSE_NUMBER",
    "Hospital ID": "HOSPITAL_ID",
    "Doctor ID": "UNIQUE_ID",
    "Telephone No": "PHONE",
    "Hospital Contact": "PHONE",
    "Fax No": "FAX",
    "E-mail ": "EMAIL",
    "URL": "URL",
    "IP_Address": "IP_ADDRESS",
    "Vehicle ID": "VEHICLE_ID",
    "Vehicle Serial No": "VEHICLE_SERIAL",
    "License Plate": "LICENSE_PLATE",
    "Device ID": "DEVICE_ID",
    "Fingerprint": "BIOMETRIC_ID",
    "Voice_ID": "BIOMETRIC_ID",
    "Photo": "PHOTO_ID",
    "Unique_ID": "UNIQUE_ID"
}

#  HELPER FUNCTION TO FIND EXACT PHI VALUES

def find_entity_spans(text, value, label):

    entities = []

    if pd.isna(value):
        return entities

    value = str(value).strip()

    if not value or value.lower() in ["nan", "none"]:
        return entities

    # Find exact text while avoiding partial word matches
    pattern = r"(?<!\w)" + re.escape(value) + r"(?!\w)"

    for match in re.finditer(pattern, text, flags=re.IGNORECASE):

        entities.append({
            "text": match.group(),
            "start": match.start(),
            "end": match.end(),
            "label": label
        })

    return entities

# DETECT DATE AND AGE PHI

def find_date_entities(text):

    entities = []

    date_pattern = (
        r"\b(?:"
        r"\d{2}[-/]\d{2}[-/]\d{4}"
        r"|"
        r"\d{4}[-/]\d{2}[-/]\d{2}"
        r")\b"
    )

    for match in re.finditer(date_pattern, text):

        entities.append({
            "text": match.group(),
            "start": match.start(),
            "end": match.end(),
            "label": "DATE"
        })

    return entities


def find_age_entities(text, age):

    entities = []

    if pd.isna(age) or float(age) <= 89:
        return entities

    age_pattern = r"\bage\s+" + re.escape(str(int(age))) + r"\b"

    for match in re.finditer(
        age_pattern, text, flags=re.IGNORECASE
    ):

        age_start = match.start() + match.group().lower().rfind("age") + 4

        entities.append({
            "text": text[age_start:match.end()],
            "start": age_start,
            "end": match.end(),
            "label": "AGE_OVER_89"
        })

    return entities

# ANNOTATE ONE CLINICAL NOTE

def annotate_note(row):

    text = str(row["Clinical note"])
    entities = []

    # Find PHI values using structured dataset columns
    for column, label in COLUMN_LABELS.items():

        if column in row.index:

            entities.extend(
                find_entity_spans(
                    text,
                    row[column],
                    label
                )
            )

    # Find calendar dates
    entities.extend(find_date_entities(text))

    # Find ages above 89 only
    if "Patient Age" in row.index:
        entities.extend(
            find_age_entities(text, row["Patient Age"])
        )

    # Remove duplicate and overlapping annotations.
    # Keep the longest annotation when spans overlap.
    entities.sort(
        key=lambda item: (
            item["start"],
            -(item["end"] - item["start"])
        )
    )

    filtered_entities = []
    last_end = -1

    for entity in entities:

        if entity["start"] >= last_end:
            filtered_entities.append(entity)
            last_end = entity["end"]

    return filtered_entities


# ANNOTATE TRAINING NOTES
if len(training_df) != 60:
    raise ValueError("Expected exactly 60 training records.")

training_annotations = []

for _, row in training_df.iterrows():

    entities = annotate_note(row)

    training_annotations.append({
        "note_id": str(row["Note ID"]),
        "text": str(row["Clinical note"]),
        "entities": entities
    })

print("TRAINING ANNOTATION")


total_entities = 0

for item in training_annotations:

    total_entities += len(item["entities"])

    print(
        item["note_id"],
        "| Detected PHI entities:",
        len(item["entities"])
    )

print("\nTotal training notes:", len(training_annotations))
print("Total detected entities:", total_entities)


# DISPLAY SAMPLE ANNOTATION

print("\nSAMPLE")


print(
    json.dumps(
        training_annotations[0],
        indent=4
    )
)
# SAVE TRAINING ANNOTATIONS


with open(
    "PHI_Training_Annotations.jsonl",
    "w",
    encoding="utf-8"
) as file:

    for item in training_annotations:
        file.write(json.dumps(item) + "\n")



# SUMMARY

print("ANNOTATION COMPLETED")
print("Training notes annotated:", len(training_annotations))
print("Training PHI entities detected:", total_entities)
print("\nSaved files:")
print("1. PHI_Training_Annotations.jsonl")


```

Character annotations into BIO labels

```python

# CONVERT CHARACTER ANNOTATIONS INTO BIO TOKEN LABELS

from transformers import AutoTokenizer
from collections import Counter

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME,
    use_fast=True
)
if not tokenizer.is_fast:
    raise ValueError("Fast tokenizer is required for character alignment.")
# Convert one annotated note into token-level labels
def convert_to_bio(note, tokenizer):

    text = note["text"]
    entities = note["entities"]

    encoded = tokenizer(
        text,
        truncation=True,
        max_length=512,
        return_offsets_mapping=True
    )

    offsets = encoded["offset_mapping"]

    # Initially, all ordinary tokens receive O
    token_labels = ["O"] * len(offsets)

    # Special tokens receive -100 so they are ignored during training
    for index, (start, end) in enumerate(offsets):
        if start == end:
            token_labels[index] = None

    # Align each character-based entity with its tokens
    for entity in entities:

        entity_start = entity["start"]
        entity_end = entity["end"]
        entity_label = entity["label"]

        matching_tokens = []

        for index, (token_start, token_end) in enumerate(offsets):

            if token_start == token_end:
                continue

            # Check whether the token overlaps the PHI entity
            if token_start < entity_end and token_end > entity_start:
                matching_tokens.append(index)

        if not matching_tokens:
            raise ValueError(
                f"Entity could not be aligned: "
                f"{entity['text']} in note {note['note_id']}"
            )

        # Assign Beginning and Inside labels
        for position, token_index in enumerate(matching_tokens):

            if token_labels[token_index] != "O":
                raise ValueError(
                    f"Overlapping entities in note {note['note_id']}"
                )

            if position == 0:
                token_labels[token_index] = "B-" + entity_label
            else:
                token_labels[token_index] = "I-" + entity_label

    # Convert BIO labels into numerical IDs
    numerical_labels = []

    for label in token_labels:

        if label is None:
            numerical_labels.append(-100)
        else:
            numerical_labels.append(label2id[label])

    return {
        "input_ids": encoded["input_ids"],
        "attention_mask": encoded["attention_mask"],
        "labels": numerical_labels,
        "offset_mapping": offsets,
        "tokens": tokenizer.convert_ids_to_tokens(
            encoded["input_ids"]
        )
    }


# Convert all training notes
training_tokenized = [
    convert_to_bio(note, tokenizer)
    for note in training_annotations
]

# Convert all testing notes
testing_tokenized = [
    convert_to_bio(note, tokenizer)
    for note in testing_annotations
]


# Display conversion results
print("BIO TOKEN CONVERSION")
print("=" * 55)

print("Fast tokenizer:", tokenizer.is_fast)
print("Training notes converted:", len(training_tokenized))
print("Testing notes converted:", len(testing_tokenized))

print("\nToken counts:")
print(
    "Training tokens:",
    sum(len(item["input_ids"]) for item in training_tokenized)
)
print(
    "Testing tokens:",
    sum(len(item["input_ids"]) for item in testing_tokenized)
)

# Count PHI labels assigned to training tokens
training_label_counts = Counter()

for item in training_tokenized:
    for label_id in item["labels"]:
        if label_id != -100:
            training_label_counts[id2label[label_id]] += 1

print("\nTraining BIO label distribution:")
for label, count in training_label_counts.items():
    print(f"{label}: {count}")

# Display one example
sample = training_tokenized[0]

print("\nSAMPLE TOKEN-LEVEL ANNOTATION")
print("-" * 55)

for token, label_id in zip(sample["tokens"], sample["labels"]):
    label = "SPECIAL TOKEN" if label_id == -100 else id2label[label_id]
    print(f"{token:20} {label}")

print("\nCharacter-to-token conversion completed successfully.")
```

# **Fine-tune the DistilBERT model**

```python

# PHI MODEL TRAINING
import torch
from torch.utils.data import Dataset, DataLoader
from transformers import (
    AutoModelForTokenClassification,
    DataCollatorForTokenClassification
)
from torch.optim import AdamW

# Prepare PyTorch dataset
class PHIDataset(Dataset):

    def __init__(self, tokenized_data):
        self.data = tokenized_data

    def __len__(self):
        return len(self.data)

    def __getitem__(self, index):

        item = self.data[index]

        return {
            "input_ids": item["input_ids"],
            "attention_mask": item["attention_mask"],
            "labels": item["labels"]
        }


# Create training and testing datasets
train_dataset = PHIDataset(training_tokenized)
test_dataset = PHIDataset(testing_tokenized)

# Prepare dynamic padding
data_collator = DataCollatorForTokenClassification(
    tokenizer=tokenizer
)

train_loader = DataLoader(
    train_dataset,
    batch_size=4,
    shuffle=True,
    collate_fn=data_collator
)

# Load DistilBERT for token classification
model = AutoModelForTokenClassification.from_pretrained(
    MODEL_NAME,
    num_labels=len(BIO_LABELS),
    id2label=id2label,
    label2id=label2id
)

# Configure optimizer
optimizer = AdamW(
    model.parameters(),
    lr=2e-5
)

# Use GPU if available, otherwise CPU
device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

model.to(device)

# Train the model
EPOCHS = 3

print("PHI MODEL TRAINING")
print("=" * 50)
print("Model:", MODEL_NAME)
print("Training notes:", len(train_dataset))
print("Testing notes reserved for evaluation:", len(test_dataset))
print("Batch size:", 4)
print("Epochs:", EPOCHS)
print("Device:", device)

for epoch in range(EPOCHS):

    model.train()
    total_loss = 0

    for batch in train_loader:

        batch = {
            key: value.to(device)
            for key, value in batch.items()
        }

        optimizer.zero_grad()

        outputs = model(**batch)

        loss = outputs.loss
        loss.backward()

        optimizer.step()

        total_loss += loss.item()

    average_loss = total_loss / len(train_loader)

    print(
        f"Epoch {epoch + 1}/{EPOCHS} "
        f"- Training Loss: {average_loss:.4f}"
    )

# Save the fine-tuned model
MODEL_SAVE_PATH = "/content/phi_distilbert_model"

model.save_pretrained(MODEL_SAVE_PATH)
tokenizer.save_pretrained(MODEL_SAVE_PATH)

print("\nModel training completed.")
print("Saved model location:", MODEL_SAVE_PATH)
```

```python

# HYBRID PHI DETECTION
# DistilBERT, Regex Rules, Training Phrase Matching

import re
import torch
import ipaddress
from datetime import datetime


# REGEX PATTERNS FOR STRUCTURED PHI



PHI_REGEX_PATTERNS = {

    "EMAIL": r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b",

    "SSN": r"\b(?:\d{3}-\d{2}-\d{4}|XXX-XX-\d{4})\b",

    "MRN": r"\bMRN[-_]?(?:(?:SYN|VAL)[-_]?)?\d+\b",

    "HEALTH_PLAN_ID": r"\bHP[-_]?(?:(?:SYN|VAL)[-_]?)?\d+\b",

    "ACCOUNT_NUMBER": r"\bACC[-_]?(?:(?:SYN|VAL)[-_]?)?\d+\b",

    "LICENSE_NUMBER": r"\bLIC[-_]?(?:(?:SYN|VAL)[-_]?)?\d+\b",

    "HOSPITAL_ID": r"\bHSP[-_]?(?:(?:SYN|VAL)[-_]?)?\d+\b",

    "VEHICLE_ID": r"\bVEH[-_]?(?:(?:SYN|VAL)[-_]?)?\d+\b",

    "VEHICLE_SERIAL": r"\bVSN[-_]?(?:(?:SYN|VAL)[-_]?)?\d+\b",

    "LICENSE_PLATE": r"(?<![\w-])(?:SYN|VAL)-\d+(?![\w-])",

    "DEVICE_ID": r"\bDEV[-_]?(?:(?:SYN|VAL)[-_]?)?\d+\b",

    "BIOMETRIC_ID": r"\b(?:FP|VP|VOICE)[-_]?(?:(?:SYN|VAL)[-_]?)?\d+\b",

    "PHOTO_ID": r"\bPHOTO[-_]?(?:(?:SYN|VAL)[-_]?)?\d+\b",

    "UNIQUE_ID": r"\b(?:BIRTHMARK|DOC|SCAR|TATTOO|UNIQUE-ID)[-_]?(?:(?:SYN|VAL)[-_]?)?\d+\b",

    "URL": r"https?://[^\s<>\"']+",

    "IP_ADDRESS": r"\b(?:\d{1,3}\.){3}\d{1,3}\b"

}




#  DATE AND AGE PATTERNS


DATE_PATTERN = (
    r"\b(?:"
    r"\d{1,2}[-/]\d{1,2}[-/]\d{4}"
    r"|"
    r"\d{4}[-/]\d{1,2}[-/]\d{1,2}"
    r")\b"
)

AGE_PATTERN = (
    r"\b(?:age\s*[:=]?\s*)"
    r"(9\d|[1-9]\d{2})\b"
    r"|"
    r"\b(9\d|[1-9]\d{2})"
    r"\s*(?:years?\s*old|y/?o)\b"
)


# PHONE AND FAX PATTERNS WITH CONTEXT


PHONE_PATTERN = (
    r"\b(?:phone|telephone|mobile|contact)\s*"
    r"[:#]?\s*(\+?\d[\d\s().-]{6,}\d)"
)

FAX_PATTERN = (
    r"\b(?:fax|facsimile)\s*"
    r"[:#]?\s*(\+?\d[\d\s().-]{6,}\d)"
)

ZIP_PATTERN = (
    r"\b(?:ZIP(?:\s*CODE)?|postal\s*code)"
    r"\s*[:#]?\s*(\d{5,6})\b"
)


# CONTEXT-BASED NAME AND FACILITY PATTERNS

PATIENT_NAME_PATTERN = (
    r"\bPatient\s+([A-Z][a-z]+(?:\s+[A-Z][a-z]+)?)"
)
RELATIVE_NAME_PATTERN = (
    r"\bRelative:\s*([A-Z][a-z]+(?:\s+[A-Z][a-z]+)?)"
)

CLINICIAN_NAME_PATTERN = (
    r"\bDr\.?\s+([A-Z][a-z]+(?:\s+[A-Z][a-z]+)?)"
)

FACILITY_PATTERN = (
    r"\b(?:visited|at|admitted\s+to|transferred\s+to)\s+"
    r"([A-Z][A-Za-z&'-]*(?:\s+[A-Z][A-Za-z&'-]*)*"
    r"\s+(?:Hospital|Clinic|Healthcare|Center|Institute))\b"
)

#  VALIDATE IP ADDRESS


def is_valid_ip(value):

    try:
        ipaddress.ip_address(value)
        return True

    except ValueError:
        return False


#  BUILD PHI PHRASE DICTIONARY FROM TRAINING DATA


TRAINING_PHI_PHRASES = {}

for annotation in training_annotations:

    for entity in annotation["entities"]:

        value = entity["text"].strip()
        label = entity["label"]

        if not value:
            continue

        if value.isdigit() and len(value) < 5:
            continue

        TRAINING_PHI_PHRASES.setdefault(label, set()).add(value)


print("Training PHI categories:", len(TRAINING_PHI_PHRASES))

print(
    "Total training phrases:",
    sum(len(values) for values in TRAINING_PHI_PHRASES.values())
)


# HYBRID PHI DETECTION FUNCTION


def detect_phi_hybrid(text, use_training_phrases=False):

    if not isinstance(text, str):
        raise TypeError("Input must be a string.")

    detected_entities = []


    # DISTILBERT CONTEXTUAL DETECTION


    model.eval()

    encoded = tokenizer(
        text,
        return_offsets_mapping=True,
        truncation=True,
        max_length=512
    )

    offsets = encoded["offset_mapping"]

    model_inputs = {
        key: torch.tensor([value], dtype=torch.long).to(device)
        for key, value in encoded.items()
        if key != "offset_mapping"
    }

    with torch.no_grad():

        outputs = model(**model_inputs)

    predictions = torch.argmax(
        outputs.logits,
        dim=2
    )[0].cpu().tolist()

    active_entity = None

    for predicted_id, (start, end) in zip(predictions, offsets):

        if start == end:
            continue

        label = id2label[predicted_id]

        if label.startswith("B-"):

            if active_entity:
                detected_entities.append(active_entity)

            active_entity = {
                "start": start,
                "end": end,
                "label": label[2:],
                "source": "DistilBERT"
            }

        elif label.startswith("I-"):

            category = label[2:]

            if (
                active_entity
                and active_entity["label"] == category
                and active_entity["end"] == start
            ):

                active_entity["end"] = end

            else:

                if active_entity:
                    detected_entities.append(active_entity)

                active_entity = {
                    "start": start,
                    "end": end,
                    "label": category,
                    "source": "DistilBERT"
                }

        else:

            if active_entity:
                detected_entities.append(active_entity)
                active_entity = None

    if active_entity:
        detected_entities.append(active_entity)



    #  REGEX DETECTION


    for label, pattern in PHI_REGEX_PATTERNS.items():

        for match in re.finditer(pattern, text, re.IGNORECASE):

            value = match.group()

            # Remove common trailing punctuation from URLs
            if label == "URL":

                while value and value[-1] in ".,;:!?)]}":
                    value = value[:-1]

                if not value:
                    continue

                end_position = match.start() + len(value)

            else:
                end_position = match.end()

            if label == "IP_ADDRESS" and not is_valid_ip(value):
                continue

            detected_entities.append({
                "start": match.start(),
                "end": end_position,
                "label": label,
                "source": "Regex"
            })


    # PHONE AND FAX DETECTION


    for label, pattern in [
        ("PHONE", PHONE_PATTERN),
        ("FAX", FAX_PATTERN)
    ]:

        for match in re.finditer(pattern, text, re.IGNORECASE):

            start, end = match.span(1)

            detected_entities.append({
                "start": start,
                "end": end,
                "label": label,
                "source": "Rule"
            })



    #  ZIP / POSTAL CODE DETECTION

    for match in re.finditer(ZIP_PATTERN, text, re.IGNORECASE):

        start, end = match.span(1)

        detected_entities.append({
            "start": start,
            "end": end,
            "label": "LOCATION",
            "source": "Rule"
        })

            #  FULL STREET ADDRESS DETECTION

    for match in re.finditer(
        LOCATION_ADDRESS_PATTERN,
        text,
        re.IGNORECASE
    ):

        detected_entities.append({
            "start": match.start(),
            "end": match.end(),
            "label": "LOCATION",
            "source": "Rule"
        })



    # FULL STREET ADDRESS DETECTION

    for match in re.finditer(
        LOCATION_ADDRESS_PATTERN,
        text,
        re.IGNORECASE
    ):

        detected_entities.append({
            "start": match.start(),
            "end": match.end(),
            "label": "LOCATION",
            "source": "Rule"
        })


    # CITY NAME DETECTION

    for match in re.finditer(
        CITY_PATTERN,
        text,
        re.IGNORECASE
    ):

        detected_entities.append({
            "start": match.start(),
            "end": match.end(),
            "label": "LOCATION",
            "source": "Rule"
        })



    #  CALENDAR DATE DETECTION


    for match in re.finditer(DATE_PATTERN, text):

        value = match.group()

        valid_date = False

        for date_format in (
            "%d-%m-%Y",
            "%d/%m/%Y",
            "%Y-%m-%d",
            "%Y/%m/%d"
        ):

            try:
                datetime.strptime(value, date_format)
                valid_date = True
                break

            except ValueError:
                continue

        if valid_date:

            detected_entities.append({
                "start": match.start(),
                "end": match.end(),
                "label": "DATE",
                "source": "Rule"
            })

   # AGE OVER 89 DETECTION


    for match in re.finditer(AGE_PATTERN, text, re.IGNORECASE):

        age_group = match.group(1) or match.group(2)

        if age_group:
            start = match.start(1) if match.group(1) else match.start(2)
            end = start + len(age_group)

            detected_entities.append({
                "start": start,
                "end": end,
                "label": "AGE_OVER_89",
                "source": "Rule"
            })


    # CONTEXT-BASED NAME AND FACILITY DETECTION

    for label, pattern in [
        ("PATIENT_NAME", PATIENT_NAME_PATTERN),
        ("RELATIVE_NAME", RELATIVE_NAME_PATTERN),
        ("CLINICIAN_NAME", CLINICIAN_NAME_PATTERN),
        ("FACILITY", FACILITY_PATTERN)
    ]:

        for match in re.finditer(pattern, text):

            if label == "CLINICIAN_NAME":
                start, end = match.span()
            else:
                start, end = match.span(1)

            detected_entities.append({
                "start": start,
                "end": end,
                "label": label,
                "source": "Rule"
            })

         # TRAINING-DERIVED EXACT PHRASE MATCHING


    if use_training_phrases:

        for label, values in TRAINING_PHI_PHRASES.items():

            matched_spans = []

            for value in sorted(values, key=len, reverse=True):

                pattern = r"(?<![\w-])" + re.escape(value) + r"(?![\w-])"

                for match in re.finditer(pattern, text, re.IGNORECASE):

                    start, end = match.span()

                    # Skip a shorter match contained inside
                    # an already matched longer phrase of the same label
                    if any(
                        start >= old_start and end <= old_end
                        for old_start, old_end in matched_spans
                    ):
                        continue

                    matched_spans.append((start, end))

                    detected_entities.append({
                        "start": start,
                        "end": end,
                        "label": label,
                        "source": "Training Phrase"
                    })

        # REMOVE EXACT DUPLICATES

    unique_entities = {}

    source_priority = {
        "DistilBERT": 1,
        "Regex": 2,
        "Rule": 2,
        "Training Phrase": 2
    }




    for entity in detected_entities:

        key = (
            entity["start"],
            entity["end"],
            entity["label"]
        )

        if key not in unique_entities:
            unique_entities[key] = entity

        elif source_priority.get(entity["source"], 0) > source_priority.get(
            unique_entities[key]["source"], 0
        ):
            unique_entities[key] = entity

    detected_entities = list(unique_entities.values())


    # RESOLVE OVERLAPPING DETECTIONS

    category_priority = {
        "SSN": 10,
        "MRN": 10,
        "HEALTH_PLAN_ID": 10,
        "ACCOUNT_NUMBER": 10,
        "LICENSE_NUMBER": 10,
        "HOSPITAL_ID": 10,
        "VEHICLE_ID": 10,
        "VEHICLE_SERIAL": 10,
        "LICENSE_PLATE": 10,
        "DEVICE_ID": 10,
        "BIOMETRIC_ID": 10,
        "PHOTO_ID": 10,
        "UNIQUE_ID": 10,
        "EMAIL": 10,
        "URL": 10,
        "IP_ADDRESS": 10,
        "DATE": 10,
        "AGE_OVER_89": 10,
        "PATIENT_NAME": 8,
        "RELATIVE_NAME": 8,
        "CLINICIAN_NAME": 8,
        "FACILITY": 8,
        "LOCATION": 7,
        "PHONE": 6,
        "FAX": 6
    }

    detected_entities.sort(
        key=lambda entity: (
            source_priority.get(entity["source"], 0),
            category_priority.get(entity["label"], 0),
            entity["end"] - entity["start"]
        ),
        reverse=True
    )



    final_entities = []

    for entity in detected_entities:

        overlaps = any(
            entity["start"] < existing["end"]
            and entity["end"] > existing["start"]
            for existing in final_entities
        )

        if not overlaps:
            final_entities.append(entity)

    final_entities.sort(
        key=lambda entity: entity["start"]
    )

    return final_entities


# FUNCTION CREATION SUMMARY


print("\nHYBRID PHI DETECTION CREATED")

print("Existing model:", MODEL_NAME)

print(
    "Model parameters:",
    f"{sum(p.numel() for p in model.parameters()):,}"
)

print("Regex categories:", len(PHI_REGEX_PATTERNS))

print("Training phrase categories:", len(TRAINING_PHI_PHRASES))

print("Detection function: detect_phi_hybrid")

print("Model retraining: Not performed")

print("Hybrid detection setup completed.")
```

# **De-identification API**

```python

# DE-IDENTIFICATION API


def deidentify(text):
    detected_entities = detect_phi_hybrid(
        text,
        use_training_phrases=False
    )

    mapping = {}
    value_to_placeholder = {}
    label_counters = {}
    all_spans = []

    # Find repeated occurrences of detected PHI values
    for entity in detected_entities:
        original_value = text[
            entity["start"]:entity["end"]
        ]

        if not original_value:
            continue

        start = 0

        while True:
            position = text.find(original_value, start)

            if position == -1:
                break

            all_spans.append({
                "start": position,
                "end": position + len(original_value),
                "value": original_value,
                "label": entity["label"]
            })

            start = position + len(original_value)

    # Remove duplicate spans and sort them
    unique_spans = {}

    for item in all_spans:
        key = (item["start"], item["end"])

        if key not in unique_spans:
            unique_spans[key] = item

    all_spans = sorted(
        unique_spans.values(),
        key=lambda item: (item["start"], -item["end"])
    )

    masked_parts = []
    last_end = 0

    for item in all_spans:
        start = item["start"]
        end = item["end"]
        original_value = item["value"]
        label = item["label"]

        # Skip overlapping spans
        if start < last_end:
            continue

        masked_parts.append(text[last_end:start])

        # Reuse placeholder for repeated values
        if original_value not in value_to_placeholder:
            label_counters[label] = (
                label_counters.get(label, 0) + 1
            )

            placeholder = (
                f"[{label}_{label_counters[label]:03d}]"
            )

            value_to_placeholder[original_value] = placeholder
            mapping[placeholder] = original_value

        else:
            placeholder = value_to_placeholder[original_value]

        masked_parts.append(placeholder)
        last_end = end

    masked_parts.append(text[last_end:])

    masked_text = "".join(masked_parts)

    return masked_text, mapping


print(" de-identify API created successfully.")
```

# **Rehydration API**

```python

# REHYDRATION API

def rehydrate(response, mapping):
    rehydrated_text = response

    for placeholder, original_value in mapping.items():
        rehydrated_text = rehydrated_text.replace(
            placeholder,
            original_value
        )

    return rehydrated_text


print("Rehydrate API created successfully.")
```

```python

# FINAL DEVELOPMENT EVALUATION

import re
from collections import defaultdict

total_tp = 0
total_fp = 0
total_fn = 0

category_counts = defaultdict(
    lambda: {"TP": 0, "FP": 0, "FN": 0}
)

for note in testing_annotations:

    text = note["text"]
    actual_entities = note["entities"]

    predicted_entities = detect_phi_hybrid(
        text,
        use_training_phrases=False
    )

    # Normalize adjacent LOCATION predictions
    sorted_predictions = sorted(
        predicted_entities,
        key=lambda x: (x["start"], x["end"])
    )

    normalized_predictions = []

    for entity in sorted_predictions:

        current = dict(entity)

        if normalized_predictions:
            previous = normalized_predictions[-1]

            gap = text[
                previous["end"]:current["start"]
            ]

            if (
                previous["label"] == "LOCATION"
                and current["label"] == "LOCATION"
                and re.fullmatch(r",\s*", gap)
            ):
                previous["end"] = current["end"]
                continue

        normalized_predictions.append(current)

    # Compare exact spans after normalization
    actual_set = {
        (e["start"], e["end"], e["label"])
        for e in actual_entities
    }

    predicted_set = {
        (e["start"], e["end"], e["label"])
        for e in normalized_predictions
    }

    true_positives = actual_set & predicted_set
    false_positives = predicted_set - actual_set
    false_negatives = actual_set - predicted_set

    total_tp += len(true_positives)
    total_fp += len(false_positives)
    total_fn += len(false_negatives)

    for _, _, label in true_positives:
        category_counts[label]["TP"] += 1

    for _, _, label in false_positives:
        category_counts[label]["FP"] += 1

    for _, _, label in false_negatives:
        category_counts[label]["FN"] += 1


# Overall metrics
precision = (
    total_tp / (total_tp + total_fp)
    if total_tp + total_fp else 0
)

recall = (
    total_tp / (total_tp + total_fn)
    if total_tp + total_fn else 0
)

f1 = (
    2 * precision * recall / (precision + recall)
    if precision + recall else 0
)

print("FINAL DEVELOPMENT EVALUATION")
print("Testing notes:", len(testing_annotations))
print("True positives:", total_tp)
print("False positives:", total_fp)
print("False negatives:", total_fn)

print(f"Precision: {precision:.4f}")
print(f"Recall:    {recall:.4f}")
print(f"F1-Score:  {f1:.4f}")

print("\nCATEGORY-WISE RESULTS")

for label in sorted(category_counts):

    counts = category_counts[label]

    tp = counts["TP"]
    fp = counts["FP"]
    fn = counts["FN"]

    p = tp / (tp + fp) if tp + fp else 0
    r = tp / (tp + fn) if tp + fn else 0
    score = (
        2 * p * r / (p + r)
        if p + r else 0
    )

    print(
        f"{label}: TP={tp}, FP={fp}, FN={fn}, "
        f"Precision={p:.4f}, Recall={r:.4f}, F1={score:.4f}"
    )
```
